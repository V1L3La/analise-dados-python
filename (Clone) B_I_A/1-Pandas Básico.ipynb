```python
import pandas as pd

url = "https://raw.githubusercontent.com/V1L3La/analise-dados-python/refs/heads/main/city_pairs.csv"

df = pd.read_csv(url)
df.head()
```

```python
df.describe(percentiles=[0.1, 0.25, 0.5, 0.75, 0.9])
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Criar o boxplot
plt.figure(figsize=(12, 6))
plt.boxplot(df['Passengers_In'], vert=False, patch_artist=True, boxprops=dict(facecolor='lightblue'))

# Mostrar o gráfico
plt.tight_layout()
plt.show()
```

```python
# Renomeando colunas
dfn = df.rename(columns={
    'Month': 'Mês',
    'AustralianPort': 'PortoAustraliano',
    'ForeignPort': 'PortoEstrangeiro',
    'Country': 'País',
    'Passengers_In': 'PassageirosEntrada',
    'Freight_In_(tonnes)': 'FreteEntrada_(toneladas)',
    'Mail_In_(tonnes)': 'CorreioEntrada_(toneladas)',
    'Passengers_Out': 'PassageirosSaída',
    'Freight_Out_(tonnes)': 'FreteSaída_(toneladas)',
    'Mail_Out_(tonnes)': 'CorreioSaída_(toneladas)',
    'Passengers_Total': 'PassageirosTotal',
    'Freight_Total_(tonnes)': 'FreteTotal_(toneladas)',
    'Mail_Total_(tonnes)': 'CorreioTotal_(toneladas)',
    'Year': 'Ano',
    'Month_num': 'NúmeroDoMês'
})
dfn.head()
```

Filtrar Passageiros Entrada maior que 3000

```python
# Filtrar clientes com renda maior que 3000
clientes_filtrados = dfn[dfn['PassageirosEntrada'] > 3000]
clientes_filtrados
```

Calcule a média de PassageirosTotal por PortoEstrangeiro.

```python
media_porto_por_passageiros = dfn.groupby('PortoEstrangeiro')['PassageirosTotal'].mean()
media_porto_por_passageiros.head(50)
```

##### Crie uma coluna “Status” com “Muito alto” se status > 5000, senão “Baixo”.

```python
dfn['Status'] = dfn['PassageirosTotal'].apply(lambda x: 'Muito alto' if x > 5000 else 'Baixo')
dfn
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 🔹 Criando dados fictícios (ex: idades de 100 pessoas)
np.random.seed(42)
idades = np.random.normal(35, 10, 100)  # Média 35, desvio padrão 10
idades = np.clip(idades, 18, 80)  # Limita entre 18 e 80 anos

# 🔹 Calcular os percentis
percentis = [0.1, 0.25, 0.5, 0.75, 0.9]
valores_percentis = np.percentile(idades, [10, 25, 50, 75, 90])

print("Valores dos percentis:")
for p, v in zip(percentis, valores_percentis):
    print(f"   {p*100:.0f}% → {v:.1f} anos")

# 🔹 Criar o gráfico de boxplot
plt.figure(figsize=(10, 6))
plt.boxplot(idades, vert=False, patch_artist=True, boxprops=dict(facecolor='lightblue'))

# 🔹 Adicionar os percentis como pontos
plt.scatter(valores_percentis, [1] * len(valores_percentis), color='red', s=100, label='Percentis')

# 🔹 Etiquetas e título
plt.xlabel('Idade (anos)')
plt.title('Distribuição de Idades com Percentis (10%, 25%, 50%, 75%, 90%)')
plt.yticks([])  # Remove rótulos do eixo Y
plt.grid(axis='x', alpha=0.3)

# 🔹 Legenda
plt.legend()

# 🔹 Mostrar o gráfico
plt.show()
```
