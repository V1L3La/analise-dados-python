```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[:2])  # Imprime os dois primeiros elementos

# Acesso por índice único        Índices começam em 0, mas podem ser negativos (-1, -2, etc.). 
print(arr[0])   # Primeiro elemento: 10
print(arr[2])   # Terceiro elemento: 30
print(arr[-1])  # Último elemento: 50
print(arr[-2])  # Penúltimo: 40

# Acesso por slicing (fatia)         Sintaxe: arr[início:fim:passo] 
print(arr[1:4])    # [20, 30, 40]  (índices 1, 2, 3)
print(arr[:3])     # [10, 20, 30]  (do início até o índice 3)
print(arr[2:])     # [30, 40, 50]  (do índice 2 até o fim)
print(arr[::2])    # [10, 30, 50]  (todos os elementos de 2 em 2)
print(arr[::-1])   # [50, 40, 30, 20, 10] (inversão)

# Acesso com lista de índices
indices = [0, 2, 4]
print(arr[indices])  # [10, 30, 50]

# Acesso com array booleano (filtro)
mask = arr > 30
print(arr[mask])  # [40, 50]
# ou
print(arr[arr > 30])  # [40, 50]
# ou
print(arr[(arr > 20) & (arr < 50)])  # [30, 40]

# Acesso com função where
posicoes = np.where(arr > 30)
print(posicoes)  # (array([3, 4]),) → Indentifica os índices 3 e 4
```

```python
import numpy as np

# Criando o array
a = np.array([3, 6, 9])
print("Array original:")
print(a)
print()

# ✅ 1. Condições lógicas simples
print("1. Condições lógicas simples:")
print("   - a > 5:", a > 5)           # [False  True  True]
print("   - a < 7:", a < 7)           # [ True  True False]
print("   - a == 6:", a == 6)         # [False  True False]
print("   - a != 6:", a != 6)         # [ True False  True]
print("   - a >= 6:", a >= 6)         # [False  True  True]
print()

# ✅ 2. Máscaras booleanas (arrays de True/False)
print("2. Máscara booleana:")
mascara = a > 5
print("   - Máscara (a > 5):", mascara)
print()

# ✅ 3. Usar máscara para filtrar dados
print("3. Filtrar elementos usando a máscara:")
print("   - Elementos onde a > 5:", a[mascara])
print("   - Ou diretamente:", a[a > 5])
print()

# ✅ 4. Operações lógicas entre máscaras
print("4. Operações lógicas entre máscaras:")
mascara1 = a > 5        # [False  True  True]
mascara2 = a < 8        # [ True  True False]

print("   - AND (e):", mascara1 & mascara2)     # [False  True False]
print("   - OR (ou):", mascara1 | mascara2)     # [ True  True  True]
print("   - NOT (não):", ~mascara1)             # [ True False False]
print()

# ✅ 5. Condições compostas com múltiplas regras
print("5. Condições compostas:")
print("   - Elementos entre 5 e 8:", a[(a > 5) & (a < 8)])
print("   - Elementos diferentes de 6:", a[a != 6])
print()

# ✅ 6. Verificar se algum ou todos os valores atendem à condição
print("6. Verificação geral:")
print("   - Alguns valores > 5?", np.any(a > 5))      # True
print("   - Todos os valores > 5?", np.all(a > 5))    # False
print()

# ✅ 7. Contar quantos elementos atendem à condição
print("7. Contagem de elementos que satisfazem a condição:")
print("   - Quantos elementos > 5?", np.sum(a > 5))    # 2
print("   - Quantos elementos == 6?", np.sum(a == 6))  # 1
```

```python
# -*- coding: utf-8 -*-
"""
Operações de Agregação em NumPy para Análise de Dados
Este script demonstra todas as funções de agregação mais usadas no NumPy,
ideal para análise exploratória de dados (EDA).
"""

import numpy as np

# ========================================
# 1. Criação de um array de exemplo
# ========================================
print("="*60)
print("1. CRIAÇÃO DE UM ARRAY DE EXEMPLO")
print("="*60)

# Criando um array 2D com dados numéricos representando vendas diárias de 3 produtos por 5 dias
vendas = np.array([
    [120, 130, 110],   # Produto A
    [95, 100, 105],    # Produto B
    [80, 85, 78]       # Produto C
])

print("Array de Vendas (3 produtos x 5 dias):")
print(vendas)
print(f"Forma do array: {vendas.shape}")
print(f"Tipo de dado: {vendas.dtype}\n")

# ========================================
# 2. Operações de Agregação Básicas
# ========================================
print("="*60)
print("2. OPERAÇÕES DE AGREGAÇÃO BÁSICAS")
print("="*60)

# 🔹 Soma total de todos os elementos
soma_total = np.sum(vendas)
print(f"Soma total de todas as vendas: {soma_total}")

# 🔹 Soma por coluna (por dia)
soma_por_dia = np.sum(vendas, axis=0)
print(f"Soma por dia (colunas): {soma_por_dia}")

# 🔹 Soma por linha (por produto)
soma_por_produto = np.sum(vendas, axis=1)
print(f"Soma por produto (linhas): {soma_por_produto}\n")

# 🔹 Média geral
media_geral = np.mean(vendas)
print(f"Média geral das vendas: {media_geral:.2f}")

# 🔹 Média por dia (colunas)
media_por_dia = np.mean(vendas, axis=0)
print(f"Média por dia: {media_por_dia}")

# 🔹 Média por produto (linhas)
media_por_produto = np.mean(vendas, axis=1)
print(f"Média por produto: {media_por_produto}\n")

# 🔹 Valor máximo absoluto
maximo = np.max(vendas)
print(f"Máximo valor em todo o array: {maximo}")

# 🔹 Valor mínimo absoluto
minimo = np.min(vendas)
print(f"Mínimo valor em todo o array: {minimo}\n")

# 🔹 Máximo por dia (colunas)
max_por_dia = np.max(vendas, axis=0)
print(f"Máximo por dia: {max_por_dia}")

# 🔹 Mínimo por dia (colunas)
min_por_dia = np.min(vendas, axis=0)
print(f"Mínimo por dia: {min_por_dia}\n")

# 🔹 Mediana geral
mediana_geral = np.median(vendas)
print(f"Mediana geral: {mediana_geral:.2f}")

# 🔹 Mediana por produto (linhas)
mediana_por_produto = np.median(vendas, axis=1)
print(f"Mediana por produto: {mediana_por_produto}\n")

# ========================================
# 3. Operações de Agregação Avançadas
# ========================================
print("="*60)
print("3. OPERAÇÕES DE AGREGAÇÃO AVANÇADAS")
print("="*60)

# 🔹 Variância geral
variancia_geral = np.var(vendas)
print(f"Variância geral: {variancia_geral:.2f}")

# 🔹 Desvio padrão geral
desvio_padrao = np.std(vendas)
print(f"Desvio padrão geral: {desvio_padrao:.2f}\n")

# 🔹 Percentis
p25 = np.percentile(vendas, 25)     # Quartil inferior
p50 = np.percentile(vendas, 50)     # Mediana
p75 = np.percentile(vendas, 75)     # Quartil superior

print(f"Percentil 25 (Q1): {p25:.2f}")
print(f"Percentil 50 (Mediana): {p50:.2f}")
print(f"Percentil 75 (Q3): {p75:.2f}\n")

# 🔹 Correlação entre produtos (linha a linha)
correlacao_produtos = np.corrcoef(vendas)
print("Matriz de correlação entre produtos:")
print(correlacao_produtos)

# 🔹 Contagem de elementos não nulos (útil para dados faltantes)
contagem = np.count_nonzero(vendas)
print(f"Contagem de valores não nulos: {contagem}\n")

# 🔹 Produto acumulado (cumulative product)
produto_acumulado = np.prod(vendas)
print(f"Produto total de todos os elementos: {produto_acumulado}\n")

# ========================================
# 4. Agregação com Eixos (axis)
# ========================================
print("="*60)
print("4. AGREGAÇÃO COM EIXOS (axis)")
print("="*60)

# Eixo 0 = colunas (vertical), Eixo 1 = linhas (horizontal)

# 🟢 Soma por coluna (eixo 0) → por dia
soma_colunas = np.sum(vendas, axis=0)
print(f"Soma por coluna (eixo 0 - por dia): {soma_colunas}")

# 🟡 Soma por linha (eixo 1) → por produto
soma_linhas = np.sum(vendas, axis=1)
print(f"Soma por linha (eixo 1 - por produto): {soma_linhas}\n")

# 🟣 Média por coluna (eixo 0)
media_colunas = np.mean(vendas, axis=0)
print(f"Média por coluna (eixo 0): {media_colunas}")

# 🟤 Média por linha (eixo 1)
media_linhas = np.mean(vendas, axis=1)
print(f"Média por linha (eixo 1): {media_linhas}\n")

# ========================================
# 5. Operações com Valores Faltantes (NaN)
# ========================================
print("="*60)
print("5. TRATAMENTO DE VALORES FALTANTES (NaN)")
print("="*60)

# Criando um array com NaN
vendas_com_nan = np.array([
    [120, np.nan, 110],
    [95, 100, np.nan],
    [80, 85, 78]
])

print("Array com valores faltantes (NaN):")
print(vendas_com_nan)

# 🔹 Ignorar NaN nas operações
soma_sem_nan = np.nansum(vendas_com_nan)
media_sem_nan = np.nanmean(vendas_com_nan)
max_sem_nan = np.nanmax(vendas_com_nan)
min_sem_nan = np.nanmin(vendas_com_nan)

print(f"Soma sem NaN: {soma_sem_nan}")
print(f"Média sem NaN: {media_sem_nan:.2f}")
print(f"Máximo sem NaN: {max_sem_nan}")
print(f"Mínimo sem NaN: {min_sem_nan}\n")

# 🔹 Verificar presença de NaN
tem_nan = np.isnan(vendas_com_nan).any()
print(f"Tem valores NaN? {tem_nan}")

# 🔹 Remover NaN (filtrar)
dados_validos = vendas_com_nan[~np.isnan(vendas_com_nan)]
print(f"Dados válidos (sem NaN): {dados_validos}\n")

# ========================================
# 6. Funções Úteis para EDA (Exploratory Data Analysis)
# ========================================
print("="*60)
print("6. FUNÇÕES ÚTEIS PARA ANÁLISE EXPLORATÓRIA DE DADOS (EDA)")
print("="*60)

# 🔹 Resumo estatístico completo com uma função
def resumo_estatistico(arr):
    print("\n📊 RESUMO ESTATÍSTICO DO ARRAY:")
    print(f"• Forma: {arr.shape}")
    print(f"• Tipo: {arr.dtype}")
    print(f"• Tamanho total: {arr.size}")
    print(f"• Soma: {np.sum(arr)}")
    print(f"• Média: {np.mean(arr):.2f}")
    print(f"• Mediana: {np.median(arr):.2f}")
    print(f"• Desvio Padrão: {np.std(arr):.2f}")
    print(f"• Variância: {np.var(arr):.2f}")
    print(f"• Máximo: {np.max(arr)}")
    print(f"• Mínimo: {np.min(arr)}")
    print(f"• Percentil 25: {np.percentile(arr, 25):.2f}")
    print(f"• Percentil 75: {np.percentile(arr, 75):.2f}")

# Aplicando ao array original
resumo_estatistico(vendas)

# ========================================
# 7. Conclusão
# ========================================
print("="*60)
print("✅ CONCLUSÃO")
print("="*60)

print("""
NumPy oferece um conjunto poderoso e eficiente de funções de agregação para análise de dados, essenciais para:
- Calcular métricas descritivas (média, mediana, desvio padrão)
- Identificar padrões e outliers
- Preparar dados para visualização e modelagem
- Tratar dados faltantes com segurança

Essas operações são rápidas, escaláveis e ideais para grandes volumes de dados — sendo a base para qualquer análise profunda com Python.
""")
```

```python
import numpy as np

print("✅ TRATAMENTO DE DADOS FALTANTES (NaN) EM NUMPY")
print("=" * 60)

# 📌 Criando um array com dados faltantes (NaN)
arr = np.array([
    [1.0, 2.0, np.nan],
    [4.0, np.nan, 6.0],
    [7.0, 8.0, 9.0],
    [np.nan, 11.0, 12.0]
])
print("Array original com NaN:")
print(arr)
print()

# ✅ 1. Detectar valores faltantes
print("1. Detectar NaN:")
mascara_nan = np.isnan(arr)
print("   Máscara de NaN:", mascara_nan)
print("   Total de NaN:", np.sum(mascara_nan))
print()

# ✅ 2. Contar quantos NaN por linha e coluna
print("2. Contagem de NaN por linha e coluna:")
print("   Por linha (soma por eixo 1):", np.sum(mascara_nan, axis=1))
print("   Por coluna (soma por eixo 0):", np.sum(mascara_nan, axis=0))
print()

# ✅ 3. Filtrar linhas sem NaN (remover linhas com dados faltantes)
print("3. Remover linhas com NaN:")
linhas_sem_nan = arr[~np.isnan(arr).any(axis=1)]
print("   Linhas sem NaN:\n", linhas_sem_nan)
print()

# ✅ 4. Filtrar colunas sem NaN
print("4. Remover colunas com NaN:")
colunas_sem_nan = arr[:, ~np.isnan(arr).any(axis=0)]
print("   Colunas sem NaN:\n", colunas_sem_nan)
print()

# ✅ 5. Preencher NaN com valor específico
print("5. Preencher NaN com valor fixo (ex: 0):")
arr_preenchido = np.where(np.isnan(arr), 0, arr)
print("   Após preenchimento:\n", arr_preenchido)
print()

# ✅ 6. Preencher NaN com média da coluna
print("6. Preencher NaN com média da coluna:")
arr_media = arr.copy()
for i in range(arr.shape[1]):
    coluna = arr[:, i]
    media = np.mean(coluna[~np.isnan(coluna)])
    arr_media[:, i] = np.where(np.isnan(coluna), media, coluna)
print("   Após preenchimento com média:\n", arr_media)
print()

```

```python
import numpy as np

# 🔹 Passo 1: Criar array 3x4 com valores aleatórios entre 1 e 100
np.random.seed(42)  # Para resultados reprodutíveis (opcional)
arr = np.random.randint(1, 101, size=(3, 4))

print("Array original (3x4):")
print(arr)
print()

# 🔹 Passo 2: Substituir valores pares por 0
# Cria uma máscara booleana onde True é par
mascara_par = (arr % 2 == 0)
arr_final = np.where(mascara_par, 0, arr)

print("Array após substituir pares por 0:")
print(arr_final)
print()

# 🔹 Passo 3: Calcular a média da matriz final
media = np.mean(arr_final)

print(f"✅ Média da matriz final: {media:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Criando uma matriz 10x10 com zeros (fundo preto)
img = np.zeros((10, 10))

# Desenhando um coração com 1s
heart_coords = [
    (1,3), (1,4), (1,5), (1,6),
    (2,2), (2,7),
    (3,1), (3,8),
    (4,1), (4,8),
    (5,2), (5,7),
    (6,3), (6,6),
    (7,4), (7,5)
]

for y, x in heart_coords:
    img[y, x] = 1

# Exibir imagem com colormap 'Reds'
plt.imshow(img, cmap='Reds')
plt.title("❤️ Desenho com NumPy")
plt.axis('off')  # Desativa os eixos
plt.show()
```

```python
salarios = [2500, 3000, 3200, 2800, 2700, 2900, 3100, 3300, 2600, 3400]

print("Média:", np.mean(salarios))  # Média salarial
print("Mediana:", np.median(salarios))  # Mediana salarial
print("Moda:", stats.mode(salarios))  # Moda salarial

print("Desvio Padrão:", np.std(salarios))  # Dispersão dos dados
print("Variância:", np.var(salarios))  # Dispersão ao quadrado

print("Máximo:", np.max(salarios))  # Maior salário
print("Mínimo:", np.min(salarios))  # Menor salário

print("Percentil 25:", np.percentile(salarios, 25))  # Quartil inferior
print("Percentil 75:", np.percentile(salarios, 75))  # Quartil superior

print(stats.describe(salarios))  # Resumo completo dos dados
```
