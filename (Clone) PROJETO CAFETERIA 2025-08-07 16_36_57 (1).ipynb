> ##### Tratamento de valores ERROR, UNKNOWN e Não informado 

```python
%python
# =====================================================================
# TRATAMENTO DE VALORES INVÁLIDOS E AUSENTES
# =====================================================================
# 1. Tratar colunas categóricas
print("🔄 Tratando colunas categóricas...")
colunas_categoricas = ['Payment Method', 'Location', 'Item']
for coluna in colunas_categoricas:
    # Substituir valores específicos por "Não informado"
    dadosloja_df[coluna] = dadosloja_df[coluna].replace(['ERROR', 'UNKNOWN'], 'Não informado')
    # Preencher valores NaN por "Não informado"
    dadosloja_df[coluna] = dadosloja_df[coluna].fillna('Não informado')
    print(f"✅ Coluna '{coluna}' tratada")

# 2. Tratar colunas numéricas (substituir por 0)
print("\n🔄 Tratando colunas numéricas...")
colunas_numericas = ['Quantity', 'Price Per Unit', 'Total Spent']
for coluna in colunas_numericas:
    # Substituir valores específicos por 0
    dadosloja_df[coluna] = dadosloja_df[coluna].replace(['ERROR', 'UNKNOWN'], 0)
    # Preencher valores NaN por 0
    dadosloja_df[coluna] = dadosloja_df[coluna].fillna(0)
    print(f"✅ Coluna '{coluna}' tratada")

# 3. Tratar colunas de data
print("\n🔄 Tratando coluna de data...")
coluna_data = ['Transaction Date']
for coluna in coluna_data:
    # Substituir valores específicos por "Não informado"
    dadosloja_df[coluna] = dadosloja_df[coluna].replace(['ERROR', 'UNKNOWN'], '2023-01-01')
    # Preencher valores NaN por "Não informado"
    dadosloja_df[coluna] = dadosloja_df[coluna].fillna('2023-01-01')
    print(f"✅ Coluna '{coluna}' tratada")

# =====================================================================
# VERIFICAÇÕES FINAIS
# =====================================================================
# 4. Informações gerais sobre o DataFrame
print("\n📋 Informações gerais do DataFrame:")
dadosloja_df.info()

# 5. Verificar valores nulos restantes
print("\n🔍 Valores nulos após tratamento:")
print(dadosloja_df.isnull().sum())

# 6. Identificar valores únicos de várias colunas separadamente
todas_colunas = ['Payment Method', 'Location', 'Item', 'Quantity', 'Price Per Unit', 'Total Spent', 'Transaction Date']
for cada_coluna in todas_colunas:
    print(f"\n=== Valores únicos na coluna '{cada_coluna}' ===")
    print(dadosloja_df[cada_coluna].unique())
```

##### Trata dados numéricos das colunas Quantity, Price Per Unit com valores zero

```python
# Tratando dados da coluna Quantity e Price Per Unit que tem valores zero

dadosloja_2tratazeros_df = dadosloja_1tiranaoinfo_df[
    (dadosloja_1tiranaoinfo_df['Quantity'] != 0) 
]

dadosloja_3tratazeros_df = dadosloja_2tratazeros_df[
    (dadosloja_2tratazeros_df['Price Per Unit'] != 0)
]

display(dadosloja_3tratazeros_df)
```

```python
# =====================================================================
# TRATAMENTO DE VALORES DE 'Total Spent' PARA ITENS ESPECÍFICOS
# =====================================================================

# Definindo os preços unitários corretos para cada item
precos_unitarios = {
    'Cake': 3,
    'Coffee': 2,
    'Cookie': 1,
    'Salad': 5,
    'Smoothie': 4,
    'Sandwich': 4,
    'Juice': 3,
    'Tea': 1.5
}

# Função para corrigir o Total Spent conforme regras fornecidas
def corrige_total_spent(row):
    item = row['Item']
    quantity = row['Quantity']
    price_per_unit = row['Price Per Unit']
    # Se o item está na lista e o preço unitário está correto
    if item in precos_unitarios and price_per_unit == precos_unitarios[item]:
        return quantity * price_per_unit
    # Se o item está na lista mas o preço unitário está incorreto, corrige pelo preço padrão
    elif item in precos_unitarios:
        return quantity * precos_unitarios[item]
    # Caso contrário, mantém o valor original
    else:
        return row['Total Spent']

# Aplicando a função ao DataFrame filtrado
dadosloja_4tratazeros_df = dadosloja_3tratazeros_df.copy()
dadosloja_4tratazeros_df['Total Spent'] = dadosloja_4tratazeros_df.apply(corrige_total_spent, axis=1)

# Exibindo o DataFrame resultante
display(dadosloja_4tratazeros_df)
```
